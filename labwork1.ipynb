```python
num = int(input("Enter a number?"))
prime = True
if num < 2:
    prime = False
else:
    for i in range (2, int(num ** 0.5)):
        if num % i == 0:
            prime = False
            break
if prime:
    print(f"{num} is a prime number")
else:
    print(f"{num} is not a prime number")
```

```python
num = int(input("Enter a number?"))
sum = sum(i for i in range(1, num) if num % i == 0)
if sum == num and num > 0:
    print(f"{num} is a perfect number")
else:
    print(f"{num} is not a perfect number")
```

```python
def extract_even(l):
    return[x for x in l if x % 2 == 0]
list = input("Enter a list?")
list = [int(x) for x in list.replace(",", "").split()]
result = extract_even(list)
print(result)
```

ex9

```python
num = int(input("Enter a non-negative interger?"))
def factorial(n):
    if n < 0:
        return "Cannot calculate factorial for negative interger"
    result = 1;
    for i in range (1, n + 1):
        result += i
    return result   
print(f"Factorial of {num} is: {factorial(num)}") 
```

ex10

```python
def get_divisors(n):
    n = abs(n)
    return[i for i in range (1, n + 1) if n % i == 0]
num = int(input("Enter a number?"))
print(f"Divisors of {num} are {get_divisors(num)}")
```
